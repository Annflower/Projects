```python
df = df.drop_duplicates(subset='student_id', keep='first')
```

```python
df['start_time'] = pd.to_datetime(df['start_time'])
df['end_time'] = pd.to_datetime(df['end_time'])
```

```python
print(df['region'].value_counts())
```

```python
# === Исправление опечаток в регионах ===
region_map = {
    'мск': 'Москва',
    'Moskva': 'Москва',
    'москва': 'Москва',
    'г. Москва': 'Москва',
    'СПб': 'Санкт-Петербург',
    'спб': 'Санкт-Петербург',
    'Санкт Петербург': 'Санкт-Петербург',
    'казань': 'Казань',
    'г.Казань': 'Казань',
}

df['region'] = df['region'].replace(region_map)

print('Регионы после очистки:')
print(df['region'].value_counts())
print(f'\nВсего уникальных регионов: {df["region"].nunique()}')
```

```python
# ============================================================
# ЭТАП 1. Уже сделано: удалили дубликаты, преобразовали даты,
#          исправили опечатки регионов
# ============================================================

# === ЭТАП 2. Обработка пропусков в заданиях ===
task_ids = range(1, 13)
correct_cols = [f'Task_{i}_correct' for i in task_ids]
time_cols    = [f'Task_{i}_time_spent' for i in task_ids]
answer_cols  = [f'Task_{i}_answer' for i in task_ids]

# correct: NaN → 0 (пропущенное задание = 0 баллов)
for col in correct_cols:
    df[col] = df[col].fillna(0)

# time_spent: NaN → 0 (не тратил время = 0 секунд)
for col in time_cols:
    df[col] = df[col].fillna(0)

# answer: оставляем NaN — это сигнал «ученик не отвечал»
# (не путать «отсутствие данных» с «нулевым ответом»)

print('Пропусков в correct:', df[correct_cols].isnull().sum().sum())
print('Пропусков в time_spent:', df[time_cols].isnull().sum().sum())
print('Пропусков в answer:', df[answer_cols].isnull().sum().sum())
```

```python
print('Максимум total_score:', df['total_score'].max())
print('Минимум total_score:', df['total_score'].min())
```

```python
# === ЭТАП 3. Производные метрики ===

# 1. Длительность прохождения (в секундах)
df['duration_sec'] = (df['end_time'] - df['start_time']).dt.total_seconds()

# 2. Итоговый балл (пересчёт для проверки)
df['total_score_calc'] = df[correct_cols].sum(axis=1)

# 3. Сколько заданий пропущено
df['tasks_skipped'] = df[answer_cols].isnull().sum(axis=1)

# 4. Сколько заданий решено правильно
df['tasks_correct'] = df[correct_cols].sum(axis=1)  # = total_score_calc

# 5. Среднее время на задачу (без учёта пропущенных)
# Заменяем 0 на NaN, чтобы пропуски не занижали среднее
df['avg_task_time'] = df[time_cols].replace(0, np.nan).mean(axis=1)

# 6. Процент решаемости (0..1)
df['correct_rate'] = df['tasks_correct'] / 12

# 7. Час старта (для анализа «в какое время суток решают»)
df['start_hour'] = df['start_time'].dt.hour

# 8. День недели
df['start_weekday'] = df['start_time'].dt.day_name()

print('Новые колонки добавлены:')
print([c for c in df.columns if c not in ['student_id'] and not c.startswith('Task_')][-8:])

print('\nОбзор новых метрик:')
df[['duration_sec', 'total_score_calc', 'tasks_skipped', 'avg_task_time', 'correct_rate']].describe().round(2)
```

```python
# === ЭТАП 4. Проверка расхождений ===
df['score_diff'] = df['total_score_calc'] - df['total_score']

mismatches = df[df['score_diff'] != 0]
print(f'Расхождений между total_score и total_score_calc: {len(mismatches)}')
print(f'Доля: {len(mismatches) / len(df) * 100:.2f}%')

if len(mismatches) > 0:
    print('\nПримеры расхождений:')
    display(mismatches[['student_id', 'total_score', 'total_score_calc', 'score_diff']].head(10))
```

```python
# === ЭТАП 5. Поиск аномалий ===

print('=' * 60)
print('ПОИСК АНОМАЛИЙ')
print('=' * 60)

# --- 5.1. Слишком быстрые сессии (< 60 сек на 12 заданий) ---
fast = df[df['duration_sec'] < 60]
print(f'\n1. Слишком быстрых сессий (< 60 сек): {len(fast)}')
if len(fast) > 0:
    display(fast[['student_id', 'region', 'duration_sec', 'total_score_calc']].head())

# --- 5.2. Слишком долгие (> 2 часов) ---
long = df[df['duration_sec'] > 7200]
print(f'\n2. Слишком долгих сессий (> 2 часов): {len(long)}')

# --- 5.3. Нулевой балл, но долгое прохождение ---
zero_long = df[(df['total_score_calc'] == 0) & (df['duration_sec'] > 300)]
print(f'\n3. Нулевой балл при длительности > 5 мин: {len(zero_long)}')

# --- 5.4. Максимальный балл за короткое время ---
fast_perfect = df[(df['total_score_calc'] >= 10) & (df['duration_sec'] < 300)]
print(f'\n4. Макс. балл за < 5 мин: {len(fast_perfect)}')

# --- 5.5. Аномально долгие ответы на одно задание (> 10 мин) ---
long_task = (df[time_cols] > 600).sum().sum()
print(f'\n5. Ответов длиннее 10 минут: {long_task}')
```

## Шаг 4. Анализ сложности заданий

Для каждого из 12 заданий рассчитаны:
- доля правильных ответов,
- среднее и медианное время выполнения.

```python
# === ШАГ 4. Сложность заданий ===
task_stats = []
for i in task_ids:
    col = f'Task_{i}_correct'
    tcol = f'Task_{i}_time_spent'
    task_stats.append({
        'Task': f'Task_{i}',
        'Правильных': int(df[col].sum()),
        'Всего ответов': int((df[col] == 1).sum() + (df[col] == 0).sum()),
        'Доля правильных, %': round(df[col].mean() * 100, 1),
        'Среднее время, сек': round(df[tcol].mean(), 1),
        'Медиана времени, сек': round(df[tcol].median(), 1),
    })

task_df = pd.DataFrame(task_stats).sort_values('Доля правильных, %', ascending=False)
print('=== Сложность заданий (от лёгкого к сложному) ===')
display(task_df)

print(f'\n📊 Самое лёгкое: {task_df.iloc[0]["Task"]} — {task_df.iloc[0]["Доля правильных, %"]}%')
print(f'📊 Самое сложное: {task_df.iloc[-1]["Task"]} — {task_df.iloc[-1]["Доля правильных, %"]}%')
```

## Шаг 7. Региональные мини-отчёты

Функция `build_region_report(region)` формирует сводную таблицу и график
для выбранного региона и сохраняет их в `reports/`. Запускается по каждому
из 10 регионов.

```python
REPORTS_DIR.mkdir(parents=True, exist_ok=True)

def build_region_report(region):
    """Сводка и график по одному региону."""
    sub = df[df['region'] == region]
    if sub.empty:
        print(f'Регион {region!r} не найден')
        return None

    # Базовая сводка
    summary = {
        'Регион': region,
        'Участников': len(sub),
        'Средний балл': round(sub['total_score_calc'].mean(), 2),
        'Медиана балла': sub['total_score_calc'].median(),
        'Доля правильных, %': round(sub['correct_rate'].mean() * 100, 1),
        'Среднее время на задачу, сек': round(sub['avg_task_time'].mean(), 1),
        'Среднее число пропусков': round(sub['tasks_skipped'].mean(), 2),
    }
    summary_df = pd.DataFrame([summary]).T
    summary_df.columns = ['Значение']

    # Распределение баллов по региону
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    sc = sub['total_score_calc'].value_counts().sort_index()
    axes[0].bar(sc.index, sc.values, color="#5F93E7")
    axes[0].set_title(f'Распределение баллов — {region}', fontsize=13)
    axes[0].set_xlabel('Итоговый балл')
    axes[0].set_ylabel('Число участников')
    axes[0].grid(axis='y', alpha=0.3)

    # Топ-5 школ региона по среднему баллу (с учётом числа участников >= 5)
    schools = (sub.groupby('school_number')
               .agg(Участников=('student_id', 'count'),
                    Средний_балл=('total_score_calc', 'mean'))
               .query('Участников >= 5')
               .sort_values('Средний_балл', ascending=False)
               .head(5))
    axes[1].bar(schools.index.astype(str), schools['Средний_балл'], color="#2E6C3D")
    axes[1].set_title(f'Топ-5 школ — {region}', fontsize=13)
    axes[1].set_xlabel('Номер школы')
    axes[1].set_ylabel('Средний балл')
    axes[1].grid(axis='y', alpha=0.3)
    plt.tight_layout()

    fname = REPORTS_DIR / f'region_{region}.png'
    plt.savefig(fname, dpi=120)
    plt.close(fig)

    return summary_df, schools, fname

# Прогоняем по всем регионам
print('=== Генерация мини-отчётов по регионам ===')
for r in sorted(df['region'].unique()):
    res = build_region_report(r)
    if res is not None:
        summary_df, schools, fname = res
        print(f"\n{r}: {summary_df.loc['Участников','Значение']:.0f} уч., "
              f"средний балл {summary_df.loc['Средний балл','Значение']:.2f}")
        print(f"   -> {fname.name}")

print('\nФайлы в reports/:', len(list(REPORTS_DIR.glob('region_*.png'))))
```

```python
# Демонстрация отчёта по одному региону (например, Москва)
demo_region = 'Москва'
demo_summary, demo_schools, demo_file = build_region_report(demo_region)
print(f'=== Мини-отчёт: {demo_region} ===')
display(demo_summary)
print('Топ-5 школ:')
display(demo_schools)
print('Сохранено:', demo_file)
```
